```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
text = ['I love learning Ai', 'I love coding', 'Ai is the future']
tokenizer = Tokenizer()
tokenizer.fit_on_texts(text)
total_words = len(tokenizer.word_index)+1
print(total_words)
```

```python
input_sequences = []
for line in text:
  token_list = tokenizer.texts_to_sequences([line])[0]
  for i in range(1, len(token_list)):
    n_gram_sequence = token_list[:i+1]
    input_sequences.append(n_gram_sequence)
```

```python
max_sequences_len = max([len(x) for x in input_sequences])
input_sequences = np.array(pad_sequences(input_sequences, maxlen = max_sequences_len, padding = 'pre'))
input_sequences
```

```python
from tensorflow.keras.utils import to_categorical

X, labels = input_sequences[:,:-1], input_sequences[:,-1]
# print(X)
# labels
Y =  to_categorical(labels, num_classes = total_words)
# print(Y)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, LSTM, Dense

model = Sequential()
model.add(Input(shape=(max_sequences_len-1,)))
model.add(Embedding(total_words, 10))
model.add(LSTM(50))
model.add(Dense(total_words, activation = 'softmax'))

model.compile(loss = 'categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()
```

```python
model.fit(X, Y, epochs=100)

seed_text = 'I love'
token_list = tokenizer.texts_to_sequences([seed_text])[0]
token_list = pad_sequences([token_list], maxlen = max_sequences_len-1, padding='pre')

predicted_probs = model.predict(token_list)
predicted_word_index = np.argmax(predicted_probs, axis = 1)[0]

print(tokenizer.index_word[predicted_word_index])
```
